````python
# eer->mdoc name mapping
base_names=("123" "238" "239" "250" "287" "457" "458" "51" "525" "547")

for id in "${base_names[@]}"; do
    prefix=""
    for f in *.eer; do
        [ -e "$f" ] || continue

        IFS=_ read -r p number rest <<< "$f"

        if [ "$number" = "$id" ]; then
            prefix="$p"
            break
        fi
    done

    echo "$id -> $prefix"
done

# example:
```
123 -> NOsy01
238 -> p46
239 -> p47
250 -> p23
287 -> p26
457 -> p49
458 -> p50
51 -> p03
525 -> p52
547 -> p41
```
````
